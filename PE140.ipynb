```python
#27 Quadratic Primes
def sieveEratosthenes(num):
    isPrime = [True] * (num + 1)
    isPrime[0] = False
    isPrime[1] = False
    primes = []
    for i in range(2, (num + 1)):
        if isPrime[i]:
            primes.append(i)
            for j in range(2 * i, (num + 1), i):
                isPrime[j] = False
    return primes
    
def isPrime(num):
    if num <= 1:
        return False
    elif num == 2 or num == 3:
        return True
    if num % 2 == 0 or num % 3 == 0:
        return False
    if not(num % 6 == 1) and not(num % 6 == 5):
        return False
    i = 5
    while i * i <= num:
        if num % i == 0 or num % (i + 2) == 0:
            return False
        i += 6
    
    return True

def equation(n, a, b):
    return n ** 2 + a * n + b

b = sieveEratosthenes(1000)
del b[0]

maxA = 0
maxB = 0
maxCount = 0

for y in b:
    for x in range(-y, 1000, 2):
        c = 0
        while isPrime(equation(c, x, y)):
            c += 1
        if maxCount < c:
            maxA = x
            maxB = y
            maxCount = c
            
print(maxA * maxB)
```

```python
# 31 Coin Sums: use graph traversal (DFS) with memoization
from functools import lru_cache
valid_coins = [1, 2, 5, 10, 20, 50, 100, 200] 
@lru_cache(None)  # memoizes the function to avoid redundant calculations
def graph_search_ways(n, max_coin):
    if n == 0:
        return 1
    return sum(
        graph_search_ways(n - coin, coin) # Recursive but not loop
        if coin <= max_coin and coin <= n else 0
        for coin in valid_coins
    )

graph_search_ways(1000,200)
```

```python
memo={0:1,1:1}
def coin_sum(amount,max_coin=200):
    if amount<0:
        return 0
    elif amount in memo:
        return memo[amount]
    ans=0
    for i in [1,2,5,10,20,50,100,200]:
        if i<=max_coin: ans+=coin_sum(amount-i,i)
        else: break
    memo.update({amount:ans})
    return ans
coin_sum(200)
```

```python
# Define the 40x40 transition matrix for the Monopoly board
M = Matrix(QQ, 40, 40, sparse=True)

dice_probs = {2: 1/36, 3: 2/36, 4: 3/36, 5: 4/36, 6: 5/36, 7: 6/36,
              8: 5/36, 9: 4/36, 10: 3/36, 11: 2/36, 12: 1/36}

chance_squares = [7, 22, 36]
community_chest_squares = [2, 17, 33]
go_to_jail = 30
jail = 10

chance_moves = {
    0: 1/16,   # Go to Go (square 0)
    10: 1/16,  # Go to Jail (square 10)
    11: 1/16,  # Go to C1 (square 11)
    24: 1/16,  # Go to E3 (square 24)
    39: 1/16,  # Go to H2 (square 39)
    5: 1/16,   # Go to R1 (square 5)
    # Next railway (R2, R3, R4 depend on current square)
    # Next utility (U1, U2 depend on current square)
    # Back 3 spaces (depends on current square)
    # 7 non-movement cards: stay put or handle via dice roll
}

cc_moves = {
    0: 1/16,   # Go to Go (square 0)
    10: 1/16,  # Go to Jail (square 10)
    # 14 non-movement cards: stay put or handle via dice roll
}

def next_railway(square):
    railways = [5, 15, 25, 35]
    for r in railways:
        if r > square:
            return r
    return railways[0]  # Wrap around to R1

def next_utility(square):
    utilities = [12, 28]
    for u in utilities:
        if u > square:
            return u
    return utilities[0]  # Wrap around to U1

# Populate transition matrix
for i in range(40):
    if i == go_to_jail:
        M[i, jail] = 1
    elif i in chance_squares:
        # Chance squares: handle card transitions
        for dest, prob in chance_moves.items():
            M[i, dest] = prob
        M[i, next_railway(i)] = 1/16
        M[i, next_utility(i)] = 1/16
        M[i, (i - 3) % 40] = 1/16
        for k, prob in dice_probs.items():
            M[i, (i + k) % 40] += (7/16) * prob
    elif i in community_chest_squares:
        # Community Chest: handle card transitions
        for dest, prob in cc_moves.items():
            M[i, dest] = prob
        for k, prob in dice_probs.items():
            M[i, (i + k) % 40] += (14/16) * prob
    else:
        # Regular squares (including Jail): follow dice rolls
        for k, prob in dice_probs.items():
            M[i, (i + k) % 40] = prob

# Verify stochastic matrix (rows sum to 1)
for i in range(40):
    row_sum = sum(M[i,j] for j in range(40))
    if abs(row_sum - 1) > 1e-10:
        print(f"Warning: Row {i} sums to {row_sum}")

# Compute steady-state distribution
eigen = M.eigenvectors_left()
stationary = None
for val, vecs, mult in eigen:
    if abs(val - 1) < 1e-10:
        stationary = vecs[0]
        stationary = [x / sum(stationary) for x in stationary]  # Normalize
        break

if stationary is None:
    print("Error: No stationary distribution found")
else:
    # Sort squares by probability
    square_probs = [(prob, i) for i, prob in enumerate(stationary)]
    square_probs.sort(reverse=True)
    # Print top squares (example output, adjust as needed)
    print("Top squares by probability:")
    for prob, square in square_probs[:5]:
        print(f"Square {square}: {prob:.6f}")
```

```python
def roman_to_int(s):
    lst=list(s)
    num=0
    count=0
    for letter in lst:
        if letter=='M':
            num+=1000
            count+=1
        else:
            break
    lst=lst[count:]
    if  lst!=[] and lst[0]=='C' and ('D' in lst or 'M' in lst):
        if lst[1]=='D':
            num+=400
            lst=lst[2:]
        elif lst[1]=='M':
            num+=900
            lst=lst[2:]
    if 'D' in lst:
        num+=500
        lst=lst[1:]
    count=0
    for letter in lst:
        if letter=='C':
            num+=100
            count+=1
        else:
            break
    lst=lst[count:]
    if  lst!=[] and lst[0]=='X' and ('L' in lst or 'C' in lst):
        if lst[1]=='L':
            num+=40
            lst=lst[2:]
        elif lst[1]=='C':
            num+=90
            lst=lst[2:]
    if 'L' in lst:
        num+=50
        lst=lst[1:]
    count=0
    for letter in lst:
        if letter=='X':
            num+=10
            count+=1
        else:
            break
    lst=lst[count:]
    if lst!=[] and lst[0]=='I' and ('V' in lst or 'X' in lst):
        if lst[1]=='V':
            num+=4
            lst=lst[2:]
        elif lst[1]=='X':
            num+=9
            lst=lst[2:]
    if 'V' in lst:
        num+=5
        lst=lst[1:]
    return num+len(lst)

def int_to_roman(num):
    lst=[]
    for _ in range(num//1000):
        lst.append('M')
    num%=1000
    if num>=900:
        lst.append('CM')
        num-=900
    if num>=500:
        lst.append('D')
        num-=500
    if num>=400:
        lst.append('CD')
        num-=400
    for _ in range(num//100):
        lst.append('C')
    num%=100
    if num>=90:
        lst.append('XC')
        num-=90
    if num>=50:
        lst.append('L')
        num-=50
    if num>=40:
        lst.append('XL')
        num-=40
    for _ in range(num//10):
        lst.append('X')
    num%=10
    if num==9:
        return ''.join(lst)+'IX'
    if num>=5:
        lst.append('V')
        num-=5
    if num==4:
        return ''.join(lst)+'IV'
    for _ in range(num):
        lst.append('I')
    return ''.join(lst)
```

```python
from functools import reduce

# Mapping of roman symbols
ROMAN_MAP = {
    'I': 1, 'V': 5, 'X': 10, 'L': 50,
    'C': 100, 'D': 500, 'M': 1000
}

# --- FP roman_to_int ---
def roman_to_int(s: str) -> int:
    vals = [ROMAN_MAP[ch] for ch in s]
    # fold: if smaller before larger => subtract, else add
    return reduce(
        lambda acc, pair: acc + (-pair[0] if pair[0] < pair[1] else pair[0]),
        zip(vals, vals[1:] + [0]),
        0
    )

# --- FP int_to_roman ---
def int_to_roman(n: int) -> str:
    # descending ordered numeral list
    table = [
        (1000, "M"), (900, "CM"), (500, "D"), (400, "CD"),
        (100, "C"), (90, "XC"), (50, "L"), (40, "XL"),
        (10, "X"), (9, "IX"), (5, "V"), (4, "IV"), (1, "I")
    ]
    # expand by comprehension & join
    return "".join(
        numeral * (n // value) for value, numeral in table
        if (n := n % value) is not None
    )
```

```python
from copy import deepcopy

def box_id(i, j):
    return (i // 3) * 3 + (j // 3)

def box_coords_from_id(b):
    br, bc = divmod(b, 3)
    r0, c0 = 3*br, 3*bc
    return [(r0 + dr, c0 + dc) for dr in range(3) for dc in range(3)]

def parse_grid(sudoku):
    """Accepts sudoku as list of 9 strings or iterables of digits (chars or ints)."""
    grid = []
    for line in sudoku:
        grid.append([int(ch) for ch in line])
    return grid

def initial_candidates(grid):
    """Return grid where filled cells are ints and empty cells are sets of candidates."""
    cand = [[None]*9 for _ in range(9)]
    for i in range(9):
        for j in range(9):
            if grid[i][j] != 0:
                cand[i][j] = grid[i][j]
            else:
                used = set(grid[i][k] for k in range(9) if grid[i][k] != 0)
                used |= set(grid[k][j] for k in range(9) if grid[k][j] != 0)
                b = box_id(i, j)
                used |= set(grid[r][c] for r,c in box_coords_from_id(b) if grid[r][c] != 0)
                cand[i][j] = set(range(1,10)) - used
    return cand

def assign(cand, i, j, val):
    """Assign a definite value to cand[i][j], propagate removal from peers.
       Return False on contradiction, True if assignment done/ok.
    """
    cur = cand[i][j]
    if isinstance(cur, int):
        return cur == val
    if val not in cur:
        return False
    # set it
    cand[i][j] = val
    # remove from peers
    for k in range(9):
        if isinstance(cand[i][k], set) and val in cand[i][k]:
            cand[i][k].remove(val)
        if isinstance(cand[k][j], set) and val in cand[k][j]:
            cand[k][j].remove(val)
    for r,c in box_coords_from_id(box_id(i,j)):
        if isinstance(cand[r][c], set) and val in cand[r][c]:
            cand[r][c].remove(val)
    return True

def propagate(cand):
    """Iteratively apply naked singles and hidden singles until no change.
       Returns False if contradiction found (empty candidate set), else cand.
    """
    while True:
        changed = False

        # 1) naked singles: assign any cell with single candidate
        for i in range(9):
            for j in range(9):
                if isinstance(cand[i][j], set):
                    if len(cand[i][j]) == 0:
                        return False
                    if len(cand[i][j]) == 1:
                        val = next(iter(cand[i][j]))
                        ok = assign(cand, i, j, val)
                        if not ok:
                            return False
                        changed = True

        # 2) hidden singles in rows
        for i in range(9):
            counts = {n: [] for n in range(1,10)}
            for j in range(9):
                if isinstance(cand[i][j], set):
                    for n in cand[i][j]:
                        counts[n].append((i,j))
            for n, cells in counts.items():
                if len(cells) == 1:
                    r,c = cells[0]
                    if isinstance(cand[r][c], set):
                        ok = assign(cand, r, c, n)
                        if not ok:
                            return False
                        changed = True

        # 3) hidden singles in columns
        for j in range(9):
            counts = {n: [] for n in range(1,10)}
            for i in range(9):
                if isinstance(cand[i][j], set):
                    for n in cand[i][j]:
                        counts[n].append((i,j))
            for n, cells in counts.items():
                if len(cells) == 1:
                    r,c = cells[0]
                    if isinstance(cand[r][c], set):
                        ok = assign(cand, r, c, n)
                        if not ok:
                            return False
                        changed = True

        # 4) hidden singles in boxes
        for b in range(9):
            counts = {n: [] for n in range(1,10)}
            for r,c in box_coords_from_id(b):
                if isinstance(cand[r][c], set):
                    for n in cand[r][c]:
                        counts[n].append((r,c))
            for n, cells in counts.items():
                if len(cells) == 1:
                    r,c = cells[0]
                    if isinstance(cand[r][c], set):
                        ok = assign(cand, r, c, n)
                        if not ok:
                            return False
                        changed = True

        if not changed:
            break
    return cand

def is_solved(cand):
    return all(isinstance(cand[i][j], int) for i in range(9) for j in range(9))

def deep_solve(cand):
    """Backtracking search using propagation. Return solved grid or None."""
    cand = propagate(cand)
    if cand is False:
        return None
    if is_solved(cand):
        return cand
    # pick unfilled cell with fewest candidates
    best = None
    best_len = 10
    for i in range(9):
        for j in range(9):
            if isinstance(cand[i][j], set):
                l = len(cand[i][j])
                if l < best_len:
                    best_len = l
                    best = (i,j)
    if best is None:
        return None
    i,j = best
    for val in sorted(cand[i][j]):
        cand_copy = deepcopy(cand)
        ok = assign(cand_copy, i, j, val)
        if not ok:
            continue
        result = deep_solve(cand_copy)
        if result is not None:
            return result
    return None

def sudoku_solve(sudoku, do_backtrack=True):
    grid = parse_grid(sudoku)
    cand = initial_candidates(grid)
    solved = propagate(cand)
    if solved is False:
        return None
    if is_solved(solved):
        return solved
    if do_backtrack:
        return deep_solve(solved)
    return solved  # partial grid with candidate sets
```

#### PE 127
- [The $abc$ conjecture](https://en.wikipedia.org/wiki/Abc_conjecture) states that there is only finitely many $abc$-hits that are slightly modified.
    - The only difference is that it is $\operatorname{rad}(abc)^{1+\epsilon}<c$, where $\epsilon>0$, unlike in the problem where $\epsilon=0$.

Let $\mathcal{H}$ be the set of all valid abc-hit triplets $(a, b, c)$ satisfying the above conditions with $c < N$. The code computes the sum of $c$ for each unique $c$ that participates in at least one such triplet. 

Using an indicator function $\mathbb{I}(\cdot)$, the final computed total is:
$$ \text{Total} = \sum_{c \in C} c \cdot \mathbb{I}\left( \exists b \in B_{L_c} \text{ such that } \begin{array}{l} b < c, \\ \text{rad}(c-b)\text{rad}(b)\text{rad}(c) < c, \\ \gcd(b, c) = 1, \\ b > c - b \end{array} \right) $$

Equivalently, expressed purely in terms of the solution set $\mathcal{H}$:
$$ \text{Total} = \sum_{(a,b,c) \in \mathcal{H}} c $$

- Firstly, we have 3 numbers, $a$, $b=c-a$ and $c$.
- $gcd(a,b)=gcd(b,c)=gcd(a,c)=1$, so they have distinct prime factors.
    - This means $rad(abc)=rad(a)*rad(b)*rad(c)$
    - Also, $gcd(a,b)=gcd(b,c)=gcd(a,c)$, so we only need to check $1$ of them.
- However, due to the large input, we need to use a single loop and check quicker with a given $a$/$b$/$c$.
- We can also factorise $a,b,c$ into at least 3 primes.
    - We take the product of those primes and make sure it is smaller than $c$.
        - We can calculate $c$ from primes through taking $max(a,b,c)$.

```python
# new radical method
size=120000
time=cputime()
total=0
lst=[1]*size
for prime in prime_range(size):
    for mul in range(1,(size-1)//prime+1):
        lst[mul*prime]*=prime
radical_lst={i:lst[i] for i in range(1,size) if 2*lst[i]<i}
# since each c has a specific floor((c-1)/rad(c)), we use it as a limit to rad(b)
b_limits=set()
for c in sorted(radical_lst):
    b_limits.add(floor((c-1)/lst[c]))
b_lst={}
for b_limit in b_limits:
    new=[]
    for b in range(8,size):      # minimum b is 8
        if lst[b]<=b_limit: new.append(b)
    b_lst[b_limit]=new
tot=0
for c in radical_lst:
    cradc=floor((c-1)/lst[c])
    num=lst[c]
    for b in b_lst[cradc]:
        if b>=c: break
        if lst[c-b]*lst[b]*num<c and gcd(b,c)==1 and b>c-b:
            tot+=c
cputime(time),tot
```

#### PE 134

Let $\mathbb{P}$ be the set of prime numbers. The total sum can be expressed as:

$$ \text{Total} = \sum_{\substack{p \in \mathbb{P} \\ 5 \le p \le 10^6}} \left[ \left( \big( q(p) - p \big) \cdot \left(10^{d(p)}\right)^{-1} \pmod{q(p)} \right) \cdot 10^{d(p)} + p \right] $$

1. **$p$**: A prime number in the range $5 \le p \le 10^6$.
2. **$q(p)$**: The next prime number strictly greater than $p$. Mathematically, $q(p) = \min \{ q \in \mathbb{P} \mid q > p \}$.
3. **$d(p)$**: The number of decimal digits of $p$, which can be written as $d(p) = \lfloor \log_{10} p \rfloor + 1$.
4. **$\left(10^{d(p)}\right)^{-1} \pmod{q(p)}$**: The modular multiplicative inverse of $10^{d(p)}$ modulo $q(p)$. This is the unique integer $x \in \{0, 1, \dots, q(p)-1\}$ such that $x \cdot 10^{d(p)} \equiv 1 \pmod{q(p)}$. 
   *(Note: This inverse always exists because $q(p) \ge 7$ is prime, meaning $\gcd(10^{d(p)}, q(p)) = 1$.)*

Thus, each term in the sum represents the **smallest non-negative integer** that ends in the exact digits of $p$ and is perfectly divisible by the next prime $q(p)$.

```python
def shortest_num(p_1,p_2):
    return inverse_mod(ceil_digit(p_1),p_2)*(p_2-p_1)%p_2
def ceil_digit(p_2):
    digits=len(str(p_2))
    return 10^digits
total=0
for i in prime_range(5,10^6):
    total+=int(str(shortest_num(i,next_prime(i)))+str(i))
total
```

```python
# 139 Pythagorean Tiles
def pythagoreanTriplets(limits):
    triplets=[]
    c,m=0,2
    while c<limits:
        for n in range(1,m):
            a=m^2-n^2
            b=2*m*n
            c=m^2+n^2
            if a+b+c<limits: triplets.append((a,b,c))
            else: break
        m+=1
    return triplets
limit=10^8
triplets=list(filter(lambda x:gcd(x)==1 and not x[2]%abs(x[0]-x[1]),pythagoreanTriplets(limit)))
print(sum(limit//sum(triplet) for triplet in triplets))
triplets
```
