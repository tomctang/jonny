```sage
# Pad Thai(run in spare time)
from pwn import * # pip install pwntools
import json
HOST = "socket.cryptohack.org"
PORT = 13421
r = remote(HOST, PORT)

def json_recv():
    line = r.readline()
    return json.loads(line.decode())

def json_send(hsh):
    request = json.dumps(hsh).encode()
    r.sendline(request)

print(r.readline())
json_send({"option":"encrypt"})
ct=eval(r.readline())['ct'];print(ct)
message=b''
for j in range(48):
    for i in range(255):
        modified=ct[:-2-2*j]
        if i<16: modified+='0'
        if j!=0: modified+=hex(i)[2:]+ct[-2*j:]
        else: modified+=hex(i)[2:]
        print(i,modified)
        json_send({"option":"unpad","ct":modified})
        if eval(r.readline())['result']:
            print('huh')
            message+=xor(int(i),int(eval('0x'+modified[-2:])))
            break
    ct=modified
    print(message[::-1])
json_send({"option":"check","message":str(message[::-1])[2:-1]})
print(r.readline())
```
